# Basic Neuron

```python
import numpy as np

def perceptron(X, w, b):
    return np.dot(X, w) + b
```

```python
# Example
X = np.array([1, 2])
w = np.array([0.5, 0.3])
b = 0.1

print(perceptron(X, w, b))
```

```python
# Activation Function

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def relu(x):
    return np.maximum(0, x)

def activation(x, act_fn = 'sigmoid'):
    if act_fn == 'sigmoid':
        return sigmoid(x)
    elif act_fn == 'relu':
        return relu(x)
    elif act_fn == 'linear':
        return x
    else:
        raise Exception('Unknown activation function')
```

```python
# Example
print(sigmoid(0.5))
print(relu(0.5))
```

```python
def perceptron(X, w, b, act_fn = 'sigmoid'):
    z = np.dot(X, w) + b
    return activation(z, act_fn)
```

```python
# Example
X = np.array([1, 2])
w = np.array([0.5, 0.3])
b = 0.1

print(perceptron(X, w, b))
print(perceptron(X, w, b, 'relu'))
print(perceptron(X, w, b, 'linear'))
```

# Layer of Neurons

```python
def dense_layer(n_neurons, X, act_fn = 'sigmoid'):
    n_input = X.shape[1]
    w = np.random.normal(size=(n_input, n_neurons))
    b = np.random.normal(size=(n_neurons))
    output = perceptron(X, w, b, act_fn)
    return output, w, b
```

```python
# Example
X = np.array([[1, 2, 3, 4]])
output, w, b= dense_layer(2, X, 'sigmoid')
print(output)
print('Weights: \n', w)
print('Bias: \n', b)
```

```python
class NeuralNetwork():
    def __init__(self, X, y, n_neurons, output_act_fn = 'sigmoid', error_fn = 'mse'):
        self.X = X
        self.y = y
        self.n_neurons = n_neurons
        self.hidden_act_fn = hidden_act_fn
        self.output_act_fn = output_act_fn
        self.error_fn = error_fn

        # Initialize weights and biases for hidden layers
        self.weights = []
        self.biases = []
        n_input = X.shape[1]
        for i in range(n_hidden_layers):
            if i == 0:
                w = np.random.normal(size=(n_input, n_neurons))
            else:
                w = np.random.normal(size=(n_neurons, n_neurons))
            b = np.random.normal(size=(n_neurons))
            self.weights.append(w)
            self.biases.append(b)

        # Initialize weights and biases for output layer
        w = np.random.normal(size=(n_neurons, 1))
        b = np.random.normal(size=(1))
        self.weights.append(w)
        self.biases.append(b)
 
    def activation(self, x, act_fn):
        if act_fn == 'sigmoid':
            return 1 / (1 + np.exp(-x))
        elif act_fn == 'relu':
            return np.maximum(0, x)
        elif act_fn == 'linear':
            return x
        else:
            raise Exception('Unknown activation function')
        
    def activation_derivative(self, act_fn, x):
        if act_fn == 'sigmoid':
            return x * (1 - x)
        elif act_fn == 'relu':
            return np.where(x > 0, 1, 0)
        elif act_fn == 'linear':
            return 1
        else:
            raise Exception('Unknown activation function')

    def forward_pass(self, X):
        hidden_output = X
        hidden_outputs = []
        # Hidden layers
        for i in range(self.n_hidden_layers):
            hidden_output = np.dot(hidden_output, self.weights[i]) + self.biases[i]
            hidden_output = self.activation(hidden_output, self.hidden_act_fn)
            hidden_outputs.append(hidden_output)
        # Output layer
        output = np.dot(hidden_output, self.weights[-1]) + self.biases[-1]
        output = self.activation(output, self.output_act_fn)
        return output, hidden_outputs
    
    def error(self, y_true, y_pred):
        if self.error_fn == 'mse':
            return np.mean(np.square(y_true - y_pred))
        elif self.error_fn == 'cross_entropy':
            return -np.mean(y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred))
        else:
            raise Exception('Unknown error function')
        
    def error_derivative(self, y_true, y_pred):
        if self.error_fn == 'mse':
            return  (y_pred - y_true) / y_true.size
        elif self.error_fn == 'cross_entropy':
            return (y_pred - y_true) / (y_pred * (1 - y_pred) * y_true.size)
        else:
            raise Exception('Unknown error function')
        
    def backward_pass(self, X, y_true, y_pred, learning_rate, hidden_outputs):
        # Output layer
        d_error_d_y_pred = self.error_derivative(y_true, y_pred)
        d_y_pred_d_z = self.activation_derivative(self.output_act_fn, y_pred)
        d_z_d_w = hidden_outputs[-1].T
        d_z_d_b = np.ones(y_pred.shape[0])

        d_error_d_w = np.dot(d_z_d_w, d_error_d_y_pred * d_y_pred_d_z)
        d_error_d_b = np.dot(d_z_d_b, d_error_d_y_pred * d_y_pred_d_z)

        self.weights[-1] -= learning_rate * d_error_d_w
        self.biases[-1] -= learning_rate * d_error_d_b

        # Hidden layers
        for i in range(self.n_hidden_layers - 1, -1, -1):
            d_y_pred_d_z = self.activation_derivative(self.hidden_act_fn, hidden_outputs[i])
            if i == 0:
                d_z_d_w = X.T
            else:
                d_z_d_w = hidden_outputs[i - 1].T
            d_z_d_b = np.ones(hidden_outputs[i].shape[0])
            d_z_d_hidden = self.weights[i + 1].T

            d_error_d_hidden = np.dot(d_error_d_y_pred * d_y_pred_d_z, d_z_d_hidden.T)
            d_error_d_w = np.dot(d_z_d_w, d_error_d_hidden * d_y_pred_d_z)
            d_error_d_b = np.dot(d_z_d_b, d_error_d_hidden * d_y_pred_d_z)

            self.weights[i] -= learning_rate * d_error_d_w
            self.biases[i] -= learning_rate * d_error_d_b

        return self.weights, self.biases
       

    def train(self, X, y, learning_rate, epochs):
        for i in range(epochs):
            y_pred, hidden_outputs = self.forward_pass(X)
            loss = self.error(y, y_pred)
            self.backward_pass(X, y, y_pred, learning_rate, hidden_outputs)
            if i % 100 == 0:
                print('Epoch %d loss: %.3f' % (i, loss)) 
```
